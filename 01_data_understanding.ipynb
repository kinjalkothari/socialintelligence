Social media platform columns show missing values that logically indicate absence or inactivity, whereas missing business metrics require careful treatment to avoid bias.

```python
cols_to_drop = ['whatsapp', 'Snapchat', 'yelp', 'vimeo']
df = df.drop(columns=cols_to_drop)
print(f"Dropped columns: {cols_to_drop}")
print(f"New DataFrame shape: {df.shape}")
```

These columns were dropped as their missing percentage in dataset is 100%.

```python
social_keywords = [
    "instagram", "facebook", "twitter", "linkedin",
    "tiktok", "youtube", "pinterest"
]

social_cols = [
    col for col in df.columns
    if any(keyword in col.lower() for keyword in social_keywords)
]

social_cols
```

These columns capture brand presence across multiple social platforms. Missing values will later be treated as inactive presence (0) rather than true nulls.

**Key Observations**
1. Dataset Structure and Scale

The dataset comprises 14,288 observations and 43 initial features, representing brand-level attributes, digital presence indicators, and performance metrics. After removing columns with no informational value, the effective feature set is reduced to 39 columns, maintaining full row coverage.

2. Data Types and Structural Inconsistencies

An initial inspection of data types reveals a mixture of object, float64, and int64 features. Notably, variables such as product_page_count, variant_count, and average_product_price are stored as object types despite representing numerical quantities. This mismatch indicates the need for explicit type coercion and validation prior to modeling, as untreated object types would prevent direct numerical analysis and model ingestion.

3. Missing Data Characteristics

Four social platform columns—WhatsApp, Snapchat, Yelp, and Vimeo—exhibit 100% missing values and were therefore removed due to the absence of any informational content.

Beyond these, a substantial number of features display high missingness, particularly:

Social media presence indicators (e.g., X, LinkedIn, YouTube, Pinterest, Twitter, TikTok)

Contact and organizational metadata (e.g., telephones, emails, employees, state)

Accessibility-related metrics (e.g., total_affected_elements, severity_critical)

Missingness across these variables ranges from moderate levels (~4–10%) to extreme cases exceeding 90–99%, suggesting that:

Certain attributes are selectively available rather than universally collected.

NaN values in social platform columns likely encode absence or inactivity, rather than data corruption.

4. Target Variable: estimated_visits

The variable estimated_visits has been selected as the primary modeling target, serving as a proxy for brand visibility and user acquisition.

Key properties:

No missing values (14,288 non-null observations).

Highly right-skewed distribution, with the mean (~6,703) substantially exceeding the median (~1,394).

Large variance, with values ranging from 0 to over 1.8 million, indicating significant heterogeneity in website traffic across brands.

These characteristics suggest the presence of outliers and non-normality, which will influence downstream modeling choices.

Risks and Modeling Considerations
1. Data Type Conversion Risk

Incorrect data types for numerically meaningful columns may lead to:

Model incompatibility

Silent exclusion of features

Incorrect statistical summaries

Careful parsing and coercion will be required to ensure valid numerical representations without unintentionally discarding meaningful observations.

2. High Missingness and Information Loss

The pervasive nature of missing data introduces several challenges:

Row-wise deletion is infeasible, as it would remove a majority of observations.

Imputation for extremely sparse features (e.g., employees, x) risks introducing artificial patterns and bias.

Feature selection and missingness-aware handling will be critical to preserve signal quality.

3. Interpretation of Social Media Absence

For social platform indicators, missing values are interpreted as absence or inactivity rather than unknown values. The chosen encoding strategy must preserve this semantic meaning while avoiding spurious correlations introduced by naive imputation.

4. Bias in Accessibility Metrics

Accessibility-related variables exhibit very high missing rates, suggesting they may only be measured for a limited subset of websites. Direct inclusion of these features could bias model behavior toward this subset unless handled with care.

5. Target Skewness

The strong right skew in estimated_visits violates assumptions of many regression models. Without appropriate transformation (e.g., logarithmic scaling), models may:

Overemphasize extreme values

Produce biased error estimates

Exhibit reduced generalization performance

6. High Cardinality in Categorical Features

Several categorical variables (e.g., website, company, technology, appsInstalled, emails) are likely to have high cardinality. Naive one-hot encoding would dramatically increase dimensionality, leading to sparsity and overfitting. Alternative encoding strategies will be required in subsequent preprocessing stages.
